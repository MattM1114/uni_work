```python
import math
answer= 0
for i in range(0,2+1):
    # Calculate the Poisson probability for i occurrences
    # with a mean of 14
    answer += math.e**(-14) * (14**i)/math.factorial(i)
    print(f"P(X={i}) = {answer:.5f}")
print(f"P(X<=2) = {answer:.5f}")
# This script calculates the Poisson probability for 0, 1, and 2 occurrences
# with a mean of 14 and prints the cumulative probability P(X<=2).
```
