```python
import numpy
import matplotlib.pyplot as plt
import seaborn as sns

dice =6
rolls = 1000
expected = 3.5
catch6 = 0

```

```python
variance = 0
for i in range(rolls):
    roll = numpy.random.randint(1, dice + 1)
    variance += (roll - expected) ** 2
    
print("Variance:", variance / rolls)
```

```python
for i in range(rolls):
    roll = numpy.random.randint(1,dice+1)
    if roll == 6:
        catch6 += 1
print("Chance of rolling a 6:", catch6 / rolls)
```

```python
# so  we have 8 apples 5 red and 3 green
# the pmf 
red = 5 
green = 3
total = 8
pmf_red = red /total
pmf_green = green / total

print("Probability of red apple:", pmf_red)
print("Probability of green apple:", pmf_green)

```
