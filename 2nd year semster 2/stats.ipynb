```python
from scipy import stats
import numpy as np

n= 10
p = 0.3
k= 3

prob = stats.binom.pmf(k,n,p)
print(f"The probability of getting exactly {k} successes in {n} trials with success probability {p} is: {prob:.4f}") 
```

```python
lambda_p = 4

k= 3
pmf_value = stats.poisson.pmf(k,lambda_p)
print(f"The probability of observing exactly {k} events when the average rate is {lambda_p} is: {pmf_value:.4f}")
```

```python
mu = 100
sigma = 15
x= 120 
cdf_value = stats.norm.cdf(x, mu, sigma)
print(f"The probability that a value is less than or equal to {x} in a normal distribution with mean {mu} and standard deviation {sigma} is: {cdf_value:.4f}")
```

```python
mean = 175
sigma =8
x = 185
z = 1- stats.norm.cdf(x,mean,sigma)
print(f"The probability that a value is greater than {x} in a normal distribution with mean {mean} and standard deviation {sigma} is: {z:.4f}")
```

```python
a = stats.norm.cdf(170,mean,sigma)
b = stats.norm.cdf(180,mean,sigma)
prob_b = b - a
print(f"The probability that a value is between 170 and 180 in a normal distribution with mean {mean} and standard deviation {sigma} is: {prob_b:.4f}")
```
