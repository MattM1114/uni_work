```python
import numpy as np
import pandas as pd
from scipy import stats

obs = {"via":[142,18],"pap smear":[108,32],"hpv":[95,55]}
obs_df = pd.DataFrame(obs)
exp = {"via":[122.67,37.33],"pap smear":[107.33,32.67],'hpv':[115,35]}
exp_df = pd.DataFrame(exp)

print(f'obs  \n{obs_df}')
print(f'exp  \n{exp_df}')
```

```python
obs_totals = obs_df.sum(axis=0)
exp_totals = exp_df.sum(axis=0)
print(f'obs total \n{obs_totals}')
print(f'exp total \n{exp_totals}')
```

```python
for test in obs_df.columns:
    chi2,p = stats.chisquare(f_obs=obs_df[test],f_exp=exp_df[test])
    print(f"{test} → Chi-square = {chi2:.3f}, p = {p:.4f}")
```
