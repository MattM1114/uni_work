```python
import numpy as np
from scipy import stats

obs = [30,25,20,15,30]
exp = [24,24,24,24,24]

```

```python
chi2,p = stats.chisquare(f_obs=obs,f_exp=exp)

print(f"chi^2 ={chi2} and p = {p}")
```

```python

obs2 = [78, 38, 30, 28, 26]
exp2 = [80, 40, 30, 30, 20]

chi2_, p_ = stats.chisquare(f_obs=obs2, f_exp=exp2)
print(f"chi^2 = {chi2_} and p = {p_}")
```
