```python
tiantic = sns.load_dataset("titanic")
tiantic.head()
```

```python
sns.stripplot(x="pclass",y='age', hue='survived',dodge=True ,data=tiantic)
plt.title("survide ")
plt.show()
```

```python
sns.stripplot(x= 'sex',y='age', hue= 'pclass', dodge=True,data=tiantic )
plt.legend(loc='upper right')
plt.grid(True)
plt.title('Age and Sex by Class')
plt.show()
```

```python
sns.stripplot(x= 'sex',y='age', hue= 'survived', dodge=True,data=tiantic )
plt.legend(loc='upper right')
plt.grid(True)
plt.title('Age and Sex by Class')
plt.show()
```

```python
# Count passengers by gender and class
gender_class_counts = tiantic.groupby(['sex',"pclass"]).size().unstack()
print("Passenger counts by gender and class:")
print(gender_class_counts)

# Visualize gender distribution by class
plt.figure(figsize=(10, 6))
sns.countplot(x='sex', hue='pclass', data=tiantic)
plt.title('Number of Passengers by Gender and Class')
plt.ylabel('Count')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.legend(title='Passenger Class')
plt.show()
# Visualize survival rate by gender and class
```

```python
# Count passengers by gender
gender_counts = tiantic['sex'].value_counts()
print("Passenger counts by gender:")
print(gender_counts)

# Optional: Visualize the gender distribution
plt.figure(figsize=(8, 5))
sns.countplot(x='sex', data=tiantic)
plt.title('Number of Passengers by Gender')
plt.xlabel('Gender')
plt.ylabel('Count')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
