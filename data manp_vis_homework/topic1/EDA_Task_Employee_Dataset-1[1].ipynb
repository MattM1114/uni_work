```python
#Import the pandas libraru
import pandas as pd
```

```python
#Import csv file
employees = pd.read_csv(r"Employee.csv")
#Visualise the data set
employees.head()
```

```python
#Basic Data Understanding
# 1
grouped_employees = employees.groupby("PaymentTier")#Group the columns by the PaymentTier column
grouped_employees.count()#count() tells you how many values there are in each row
```

```python
#2
grouped_gender = employees.groupby("Gender")
grouped_gender.count()
```

```python
#3
grouped_education = employees.groupby("Education").agg({"Age":["mean"]})#Group the columns by education and then finds the average age of each level of education
grouped_education
```

```python
#4
#Groups the columns by the year the employees joined and then gives the min,max and average of experience for each year
grouped_joiningYear = employees.groupby("JoiningYear").agg({"ExperienceInCurrentDomain":["min","max","mean"]})
grouped_joiningYear
```

```python
#Intermediate Aggregations
#5
grouped_city = employees.groupby("City").agg({"EverBenched":["sum"]})#Groups by the city and adds all the values of the everbenched column
grouped_city
```

```python
#6
#Group by two columns, gender and paymentTier and then finds the average of age and experience for both male and female and then each payment tier
grouped_gender_payment = employees.groupby(["Gender","PaymentTier"]).agg({"Age":["mean"],"ExperienceInCurrentDomain":["mean"]})
grouped_gender_payment
```

```python
#7
#Groups by education and finds which education level has the highest turnover rate, since 1 means that the employee left and 0 means that they stayed
#I used max to get all the 1's for each education level and then used count to see which education level has the most amount of 1's, therefore the highest turnover rate
grouped_educationlvl = employees.groupby("Education").agg({"LeaveOrNot":["max","count"]})
grouped_educationlvl
```

```python
#8
#Grouping by two columns to find the middle most values for age and experience for both male and female and each education level
grouped_gender_education = employees.groupby(["Gender","Education"]).agg({"Age":["median"],"ExperienceInCurrentDomain":["median"]})
grouped_gender_education
```

```python
#Advanced Analysis
#9
#Comparing the three payment tiers with each level of education
#Essentially looking at the amount of employees in each education level and payment tier. Eg.How many employees have a bachelor's degree that are in payment tier 1. 
cross_tab_employees = pd.crosstab(employees["Education"],employees["PaymentTier"])
cross_tab_employees
```

```python
#10
#Groups the columns by joiningYear and finds the total of the everbenched column for each year
grouped = employees.groupby("JoiningYear").agg({"EverBenched":"sum"})
grouped
```

```python
#10
#Finds the average experience for each year
grouped2 = employees.groupby("JoiningYear").agg({"ExperienceInCurrentDomain":["mean"]})
grouped2 
```

```python
#11
#Groups by city and education level and finds the minimum and maximum age for each education level in that city
group = employees.groupby(["City","Education"]).agg({"Age":["min","max"]})
group
```

```python
#12
#Groups by paymentTier and finds the minimum, maximum and average values of the leaveOrNot column for each payment tier.
group2 = employees.groupby("PaymentTier").agg({"LeaveOrNot":["min","max","mean"]})
group2
```
