```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

#read the data from the csv file
df1 = pd.read_csv('Employee.csv')

#count empolyes by joiningYear
Joining_count = df1.groupby('JoiningYear').size()

plt.figure(figsize=(10,6))

plt.plot(Joining_count.index, Joining_count.values, color='blue', marker='o', linestyle='dashed', linewidth=2, markersize=12)

plt.xlabel('JoiningYear')
plt.ylabel('Number of employees')
plt.title('Number of employees by joining year')

plt.show()
```
