```python
import pandas as pd 
import plotly.express as px
import seaborn as sns

mpg = sns.load_dataset('mpg')

mpg.head()
```

```python
fig = px.scatter(mpg,
                 x='horsepower',
                 y='mpg',
                 hover_data=['name'],
                 color='origin',
                 size = 'cylinders',
                 title= 'vehicle mpg by horsepower',
                 labels={'horsepower':'Horsepower', 'mpg':'Miles per gallon',})

fig.show()

```

```python
import pandas as pd
import seaborn as sns
import plotly.express as px

p = sns.load_dataset('penguins')

fig = px.box(
    p,
    x='species',
    y='body_mass_g',
    color =  'island',
    hover_data=['sex','bill_length_mm','flipper_length_mm','body_mass_g'],
    title='Penguin characteristics by species',
)
fig.show()
```
