```python
import numpy as np

temperature = np.array([[23, 25, 28, 30], [21, 22, 26, 29], [24, 27, 29, 32]])
humidity = np.array([[60, 65, 70, 75], [55, 58, 68, 72], [62, 67, 74, 78]])
```

```python
mean_temperature = np.mean(temperature)
mean_humidity = np.mean(humidity)
print(f"Mean Temperature: {round(mean_temperature,2)}, \n Mean Humidity:, {round(mean_humidity,2)}")
```

```python
max_h = np.max(humidity)
print(max_h)
```

```python
comfort_index = 0.8*temperature*0.2*humidity
print(f"comfort index is\n{comfort_index}")
```

```python
# Compute the variance of temperature along each row.
variance_temperature = np.var(temperature, axis=1)
print(variance_temperature)
```

```python
prices = np.array([[200, 150, 300], [120, 180, 250], [100, 140, 190]])
category_discounts = np.array([[5], [8], [12]])
discount = 0.10*prices
discounted_prices = prices - discount
print(discounted_prices)
```

```python
category_s = (category_discounts/100)

print(category_s)
```

```python
finalprice = prices -discount - category_s
print(finalprice)
```
