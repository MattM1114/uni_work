```python
class Vehicle:
    def __init__(self, brand, speed):
        self.brand = brand
        self.speed = speed

    def drive(self):
        return f"The {self.brand} is driving at {self.speed} km/h."
```

```python
class Car(Vehicle):
    def __init__(self, brand, speed, fuel_type):
        super().__init__(brand, speed)
        self.fuel_type = fuel_type

    def refuel(self):
        return f"The {self.brand} car is refueling with {self.fuel_type}."
```

```python
class Product:
    def __init__(self, name: str, price: float):
        """Initialize a Product with a name and price."""
        self.name = name
        self.price = price
    
    def __str__(self):
        """Return product details when printed."""
        return f"{self.name}: ${self.price:.2f}"
```

```python
class Order:
    def __init__(self):
        """Initialize an Order with an empty list of products."""
        self.products = []

    def add_product(self, product: Product):
        """Add a product to the order."""
        self.products.append(product)

    def calculate_total(self) -> float:
        """Calculate and return the total price of all products in the order."""
        return sum(product.price for product in self.products)

    def __str__(self):
        """Return order summary."""
        return f"Order contains {len(self.products)} products. Total: ${self.calculate_total():.2f}"
```

```python
# Payment class to process transactions
class Payment:
    def __init__(self, order: Order, amount: float):
        """
        Initialize a Payment with an associated order and payment amount.
        Ensures payment amount covers the order total.
        """
        self.order = order
        self.amount = amount

    def process_payment(self) -> str:
        """
        Process the payment.
        Check if the payment amount is sufficient.
        """
        total_cost = self.order.calculate_total()
        if self.amount >= total_cost:
            return f"Payment of ${self.amount:.2f} processed successfully."
        else:
            return f"Insufficient payment. Total cost is ${total_cost:.2f}."
```
