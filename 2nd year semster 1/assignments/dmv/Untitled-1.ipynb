```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(42)

```

```python
# Quick service (60%): average wait time = 15 mins
q_ser = np.random.poisson(lam=15,size=6000)

# Full service (40%): average wait time = 45 mins
full_ser =np.random.poisson(lam=45, size=3000)

# Combine population (9000 total)
population =np.concatenate([q_ser, full_ser])
```

```python
pop_mean = np.mean(population)
pop_median = np.median(population)
pop_std =np.std(population,ddof=0)
```

```python
print("=== POPULATION STATISTICS ===")
print(f"Population size: {len(population)}")
print(f"Mean: {pop_mean:.2f} minutes")
print(f"Median: {pop_median:.2f} minutes")
print(f"Standard deviation: {pop_std:.2f} minutes")
```

```python
plt.figure(figsize=(10, 6))
sns.histplot(population, bins=50, kde=False, color="green", edgecolor="black")
plt.title("Population Wait Time Distribution (Kai’s Sea Food)")
plt.xlabel("Wait Time (minutes)")
plt.ylabel("Frequency")
plt.show()
```

```python
n = 30          # sample size
num_samples = 500

sample_means = []
```

```python
for i in range(num_samples):
    sample = np.random.choice(population, size=n, replace=False)
    sample_means.append(np.mean(sample))
```

```python
sample_means = np.array(sample_means)

# Compute sampling stats
sample_mean = np.mean(sample_means)
sample_std = np.std(sample_means, ddof=1)
theoretical_se = pop_std / np.sqrt(n)
```

```python
print("\n=== SAMPLING DISTRIBUTION STATISTICS ===")
print(f"Number of samples: {num_samples}")
print(f"Sample size: {n}")
print(f"Mean of sample means: {sample_mean:.2f} minutes")
print(f"Standard deviation of sample means: {sample_std:.2f} minutes")
print(f"Theoretical Standard Error (σ/√n): {theoretical_se:.2f} minutes")
```

```python
plt.figure(figsize=(10, 6))
sns.histplot(sample_means, bins=30, kde=True, color="blue", edgecolor="black")
plt.title("Sampling Distribution of Sample Means (n=30, 500 samples)")
plt.xlabel("Sample Mean Wait Time (minutes)")
plt.ylabel("Frequency")
plt.show()
```

```python
print("\n=== CENTRAL LIMIT THEOREM DEMONSTRATION ===")
print(f"Population mean ≈ {pop_mean:.2f}")
print(f"Mean of sample means ≈ {sample_mean:.2f}")
print("✅ The sample means approximate the population mean, showing unbiasedness.")
print("✅ The sampling distribution is approximately normal even though the population is bimodal.")
print("✅ The spread (SD of means) ≈ theoretical standard error (σ/√n).")
```
