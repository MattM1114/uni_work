```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

iris = sns.load_dataset('iris')
iris.head()
```

```python
plt.figure(figsize=(12, 6))
sns.jointplot(x='sepal_length',y='petal_length',data=iris,hue='species',kind='scatter')
plt.grid()
plt.title('Sepal Length vs Petal Length')
plt.tight_layout()
plt.show()

```

so here we are observing the petale length vs the sepal length based on the species
so acording to the data we can see that the setosa has the smallest petale length and the virginica has the largest petale length
the setosa is set to blue the versicolor is orange and the virginica is green
we can see the size of the petale length is different for each species  based on its category  which is the species and were it falls on the graph 
we can see form this that sepal length and petale length have a positive correlation
