```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt

# Load the data
data = pd.read_excel('laptopPrice2.xlsx')

data.head()
```

```python
data.describe()
```

```python
data.info()
```

```python
# Check for missing values
data.isnull().sum()
```

```python
#trying to group the data by price and rating
price_rating = data.groupby(['rating'])['Price'].mean()
price_rating
```

```python
# Plot the data
plt.plot(price_rating.index, price_rating.values)
plt.legend()
plt.title('Price vs Rating')
plt.xlabel('Rating')
plt.ylabel('Price')
plt.show()
```

```python
# group the data by brand and price
brand_price = data.groupby(['brand'])['Price'].mean()
brand_price
```

```python
# Plot the data
plt.bar(brand_price.index, brand_price.values)
plt.legend()
plt.title('Brand vs Price')
plt.xlabel('Brand')
plt.ylabel('Price')
plt.show()
```

```python
core_price = data.groupby(['processor_name'])['Price'].mean()
core_price
```

```python
plt.bar(core_price.index, core_price.values)
plt.title('Processor vs Price')
plt.xlabel('Processor')
plt.ylabel('price')
plt.xticks(rotation=90)
plt.show()
```
