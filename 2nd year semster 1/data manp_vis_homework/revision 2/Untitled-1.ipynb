```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns


data = pd.read_csv("customer_data.csv")
df = pd.DataFrame(data)
# checking for null valvues
df.isnull().sum
```

```python
print(df.head())
```

```python
df['income'].fillna(df['income'].median())
```

```python
plt.figure(figsize=[10,6])
sns.scatterplot(data=df,x=df['age'],y=df['income'],hue=df['education_level'],palette='Greens')
plt.title("age vs income ")
plt.show
```

```python
plt.figure(figsize=[10,6])
heat = df.drop(['education_level', 'customer_id'], axis=1)

sns.heatmap(heat.corr(),annot=True, cmap='coolwarm', fmt=".2f")
plt.title("customer_data")
plt.show()
```

```python
p_data = sns.load_dataset("penguins")
penguins_df = pd.DataFrame(p_data)
penguins_df.head()
```

```python
plt.figure(figsize=(10,6))
sns.jointplot(data=penguins_df,x=penguins_df["bill_length_mm"],y=penguins_df['bill_depth_mm'],hue='species',)
plt.title("bill_length_mm vs bill_depth_mm")
plt.show()
```
