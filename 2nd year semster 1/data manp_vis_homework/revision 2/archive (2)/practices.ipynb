```python
import pandas as pd 
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
import plotly.express as px

data = pd.read_csv("student_habits_performance.csv")
student_df = pd.DataFrame(data)
student_df.info()
```

```python
print(student_df.dtypes)
```

```python
student_df.isnull().sum
```

```python
student_df.head()
```

```python
plt.Figure(figsize=[10,6])
sns.heatmap(student_df.select_dtypes(include=['float64','int64']).corr(),annot=True,cmap='coolwarm',fmt='.2f')
```

```python
plt.Figure(figsize=[10,6])
sns.scatterplot(data=student_df,x=student_df['study_hours_per_day'],y=student_df['exam_score'],hue=student_df['parental_education_level'])
```

```python
plt.Figure(figsize=[10,6])
sns.jointplot(data=student_df,x=student_df['social_media_hours'],y=student_df['exam_score'],hue=student_df['parental_education_level'])
```

```python
student_df.count()

```

```python
part =student_df.groupby('part_time_job').size().reset_index(name='count')
part.columns = ['part_time', 'count']

# Plot pie chart
fig = px.pie(part, values='count', names='part_time', title='Student with part time jobs')
fig.show()
```

```python
sns.jointplot(data=student_df,x=student_df['attendance_percentage'],y=student_df['exam_score'],hue=student_df['gender'])
```

```python
sns.lineplot(data=student_df,x=student_df['mental_health_rating'],y=student_df['exam_score'])
```
