```python

import matplotlib.pyplot as plt
import seaborn as sns

tips = sns.load_dataset("tips")
tips.head()
```

```python
sns.catplot(x='day',y='total_bill',kind ='violin',data= tips)
plt.title('Violin plot')
plt.tight_layout()
plt.show()
```

```python
sns.catplot(x='day',y='total_bill',kind ='bar',data= tips)
plt.title(' bar plot')
plt.tight_layout()
plt.show()
```

```python
sns.catplot(x='day',y='total_bill',kind ='point',data= tips)
plt.title(' point plot')
plt.tight_layout()
plt.show()
```

```python
sns.catplot(x='day',y='total_bill',kind ='boxen',data= tips)
plt.title(' boxen plot')
plt.tight_layout()
plt.show()
```

```python
mpg = sns.load_dataset('mpg')

mpg.head()
```

```python
plt.Figure(figsize=(10,6))
sns.displot(data =mpg, x= 'mpg',kind='hist',bins=10 ,palette='viridis')
plt.title('Histogram of mpg')
plt.xlabel('mpg')
plt.ylabel('Frequency')
plt.show()
```

```python
titianic = sns.load_dataset('titanic')
titianic.head()
```

```python

sns.catplot(x='class',y='fare', data=titianic,hue='sex',kind='box',palette='dark')
plt.title('Fare by Class ')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.catplot(x='day',y='tip',data=tips,kind='box',palette='pastel')
plt.title('Boxplot of Tips by Day')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.catplot(x='day',y='total_bill',data=tips,kind='boxen',hue='time')
plt.title('Total Bill by Day and Time')
plt.legend(title='Time')
plt.show()
```

```python
sns.relplot(x='day',
            y='total_bill',
            kind='line',
            palette='dark',
            data=tips)

plt.tight_layout()
plt.show()
```

```python
flights = sns.load_dataset("flights")
flights.head()
```

```python
sns.relplot(x='month',
            y='passengers',
            hue='year',
            palette='Blues',
            data=flights,
            kind='line')
plt.tight_layout()
plt.show()

```
