```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('murder_2015.csv')
df = pd.DataFrame(df)
df.head()
```

```python
#2. Calculate the total number of murders for each state by summing the counts from 2014 and 2015
total_murders_state = df.groupby('state')[['2014_murders','2015_murders']].sum()
total_murders_state['total_murders'] = total_murders_state['2014_murders'] + total_murders_state['2015_murders']

total_murders_state.reset_index(inplace=True)

print(total_murders_state)
```

```python
#3. Identify the top 5 states with the highest combined murders.
top_5_states = total_murders_state.groupby('state')['total_murders'].sum().nlargest(5)
print(f"Top 5 States with the Highest Combined Murders:{top_5_states}")
```

```python
plt.figure(figsize=(12, 6))
plt.scatter(total_murders_state['state'], total_murders_state['total_murders'], s=50, alpha=0.7)

for i ,row in total_murders_state.iterrows():
    plt.annotate(str(int(row['total_murders'])),
                 (row['state'],row['total_murders']),
                 textcoords="offset points",
                 xytext=(0,10),
                 ha='center')
plt.title('Total Murders by State')
plt.xlabel('State')
plt.ylabel('Total Murders')
plt.xticks(rotation=90)
plt.grid(True)
plt.tight_layout()
plt.show()
```
